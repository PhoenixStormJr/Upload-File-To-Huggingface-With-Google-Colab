```python
#@title upload file to hugging face

#@markdown get your hf token in https://huggingface.co/settings/tokens
HF_TOKEN = "" #@param {type:"string"}
#@markdown Your repo_id is the name of your username + whatever you want to name the folder where it will be saved  🤗
repo_id = "PhoenixStormJr/Megaman-NT-Warrior-Lan_s-Mother-Hakura-RVC" #@param {type:"string"}

Use_Google_Drive = True # @param {type:"boolean"}

import os
if(Use_Google_Drive):
  if(not os.path.exists("/content/drive/MyDrive")):
    from google.colab import drive
    drive.mount('/content/drive/')


if(Use_Google_Drive):
  if(not os.path.exists("/content/drive/MyDrive/UploadToHuggingface")):
    os.makedirs("/content/drive/MyDrive/UploadToHuggingface")
  if(len(os.listdir("/content/drive/MyDrive/UploadToHuggingface")) < 1):
    from google.colab import files
  while(len(os.listdir("/content/drive/MyDrive/UploadToHuggingface")) < 1):
    print("ERROR: There are no files in the folder: \"UploadToHuggingface\", please upload a file there.")
    %cd /content/drive/MyDrive/UploadToHuggingface
    uploaded = files.upload()
else:
  !rm -r /content/Uploaded
  !mkdir -p /content/Uploaded
  %cd /content/Uploaded
  from google.colab import files
  uploaded = files.upload()
  fileUploaded = list(uploaded.keys())[0]
  print(fileUploaded)





from huggingface_hub import HfApi, create_repo, login
login(HF_TOKEN)
try:
    create_repo(repo_id)
except:
    print("Error creating repo, proceeding...")


if(Use_Google_Drive):
  for fileUploaded in os.listdir("/content/drive/MyDrive/UploadToHuggingface"):
    api = HfApi()
    api.upload_file(
        path_or_fileobj=f"/content/drive/MyDrive/UploadToHuggingface/{fileUploaded}",
        path_in_repo=f"{fileUploaded}",
        repo_id=repo_id,
        repo_type="model",
    )
else:
  api = HfApi()
  api.upload_file(
      path_or_fileobj=f"/content/Uploaded/{fileUploaded}",
      path_in_repo=f"{fileUploaded}",
      repo_id=repo_id,
      repo_type="model",
  )



```
